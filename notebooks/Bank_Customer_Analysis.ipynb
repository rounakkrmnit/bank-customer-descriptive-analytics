```python
from scipy.stats import chi2_contingency
```

```python
def chi_square_test(df, feature, target="Exited"):
    contingency_table = pd.crosstab(df[feature], df[target])
    
    chi2, p_value, dof, expected = chi2_contingency(contingency_table)
    
    print(f"Feature: {feature}")
    print(f"Chi-square statistic: {chi2:.2f}")
    print(f"P-value: {p_value:.6f}")
    
    if p_value < 0.05:
        print("Result: Statistically significant association with attrition")
    else:
        print("Result: No statistically significant association with attrition")
```

```python
from scipy.stats import chi2_contingency

def cramers_v(df, feature, target="Exited"):
    table = pd.crosstab(df[feature], df[target])
    chi2, p_value, dof, expected = chi2_contingency(table)

    n = table.sum().sum()
    r, k = table.shape

    phi2 = chi2 / n
    phi2_corrected = max(0, phi2 - ((k - 1) * (r - 1)) / (n - 1))
    
    r_corrected = r - ((r - 1) ** 2) / (n - 1)
    k_corrected = k - ((k - 1) ** 2) / (n - 1)

    v = np.sqrt(
        phi2_corrected /
        min(k_corrected - 1, r_corrected - 1)
    )

    return v
```

```python
features = [
    "Geography",
    "Gender",
    "IsActiveMember",
    "NumOfProducts"
]

results = []

for feature in features:
    results.append({
        "Feature": feature,
        "Cramers_V": round(cramers_v(analysis_df, feature), 3)
    })

effect_size = pd.DataFrame(results).sort_values(
    "Cramers_V",
    ascending=False
)

effect_size
```
